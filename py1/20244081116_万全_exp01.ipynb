实验2：函数、条件、循环与递归

```python
#1 score_level
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0~100")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(70))

#2 for while求1~100和
sum_for =0
for i in range(1,101):
    sum_for +=i

sum_while=0
i=1
while i<=100:
    sum_while +=i
    i +=1
print("for求和",sum_for,"while求和",sum_while)

#3阶乘：迭代、递归
def fact_iter(n):
    res=1
    for x in range(2,n+1):
        res *=x
    return res

def fact_rec(n):
    if n ==0 or n==1:
        return 1
    return n * fact_rec(n-1)

for num in [0,1,5,10]:
    print(f"n={num}, iter={fact_iter(num)}, rec={fact_rec(num)}")

#4 summary 返回最小最大平均
def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

res = summary([10,20,30])
mi,ma,av = res
print("summary", mi,ma,av)

#5 sum_all(*args)可变参数
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1,2,3,4))
```

实验3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

#1 zip构建字典
stu_dict = dict(zip(names, scores))
for k,v in stu_dict.items():
    print(k,v)

#2 enumerate序号从1
for idx, (name,sc) in enumerate(zip(names,scores), start=1):
    print(idx, name, sc)

#3列表推导式
pass_names = [n for n,s in zip(names,scores) if s>=60]
score_sq = [s**2 for s in scores]
print("及格姓名", pass_names)
print("成绩平方",score_sq)

#4统计
count = len(scores)
avg = sum(scores)/count
maxs = max(scores)
mins = min(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / count
print(f"人数:{count},平均分{avg:.2f},最高{maxs},最低{mins},及格率{pass_rate:.1%}")

#5 列表浅拷贝
demo = [10,20,30]
alias = demo
copied = demo.copy()
alias[0] = 999
print("原demo",demo)
print("alias",alias)
print("copied副本",copied)

#6元组不可赋值；zip长短不一
t = (11,22,33)
try:
    t[0]=99
except TypeError as e:
    print("元组赋值报错",e)

list1 = [1,2]
list2 = [10,20,30,40]
print("zip不等长序列:", list(zip(list1,list2)))
```

4.
统计结果：人数为 6；最高分为 92；最低分为 58；平均分为 76.666...；及格人数为 5 人，及格率为 83.333...%
格式化：使用 f-string 格式化，平均分使用 :.2f 保留两位小数（输出 76.67）；及格率使用 :.2% 格式化为百分数（输出 83.33%）
5.
alias = scores 是引用赋值（别名），修改 alias 中的元素，原 scores 列表会跟着改变。
copied = scores.copy() 是浅拷贝，创建了一个新列表。修改 alias（即修改原列表）时，copied 列表不会发生变化。因此，原列表改变，副本保持不变
6.
元组不可变：尝试对元组内的元素重新赋值（如 t[0] = 1）会引发 TypeError，因为元组创建后其内部元素的引用不可更改。
长度不同的 zip()：当两个序列长度不同时，zip() 会以较短序列的长度为准，多余的元素会被直接截断丢弃，不会报错。例如 zip(["A", "B", "C"], [1, 2]) 的结果只有 ('A', 1), ('B', 2)



实验4：⽂件读写、异常处理与模块

```python
import os
import pickle

#1路径创建
cwd = os.getcwd()
print("当前工作目录", cwd)
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
#写入students.txt
txt_path = os.path.join(data_dir,"students.txt")
content = '''name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
'''
with open(txt_path,"w",encoding="utf-8") as f:
    f.write(content)
#4.3 parse_student函数
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError(f"字段数量错误，行内容:{line}")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数:{score_str}")
    if not (0 <= score <=100):
        raise ValueError(f"成绩越界0~100:{score}")
    return {"name":name, "score":score}
#4.4逐行处理捕获异常
valid_records = []
error_log = []
with open(txt_path,"r",encoding="utf-8") as f:
    lines = f.readlines()
for lineno, line in enumerate(lines, start=1):
    if lineno ==1:
        continue #跳过表头
    try:
        rec = parse_student(line)
        if rec is not None:
            valid_records.append(rec)
    except ValueError as err:
        error_log.append((lineno, str(err)))
print("有效记录：", valid_records)
print("错误行：", error_log)
#4.5 report.txt
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_num = sum(1 for r in valid_records if r["score"]>=60)
    pass_r = pass_num / valid_num
else:
    avg_score =0
    pass_r=0
report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg_score:.2f}\n")
    f.write(f"及格率:{pass_r:.2%}\n")
# a追加模式演示
with open(report_path,"a",encoding="utf-8") as f:
    f.write("---追加一行测试内容---\n")
#4.6 pickle读写
pkl_path = os.path.join(data_dir,"students.pkl")
with open(pkl_path,"wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path,"rb") as f:
    load_back = pickle.load(f)
print("pickle读取回来：", load_back)
#4.7 输出parse_student到外部py文件
util_code = '''
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError(f"字段数量错误，行内容:{line}")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数:{score_str}")
    if not (0 <= score <=100):
        raise ValueError(f"成绩越界0~100:{score}")
    return {"name":name, "score":score}
'''
with open("student_utils.py","w",encoding="utf-8") as f:
    f.write(util_code)
#导入测试
from student_utils import parse_student
print("外部模块测试有效行：", parse_student("Tom,77"))
try:
    parse_student("Bad,-5")
except ValueError as e:
    print("外部模块测试无效：", e)
#4.8捕获文件不存在
try:
    with open("no_exist.txt","r") as f:
        pass
except FileNotFoundError as e:
    print("捕获不存在文件异常：",e)
```


w 模式（覆盖写入）：打开文件时，会清空该文件原有的所有内容。如果 report.txt 里之前有数据，会被彻底抹除，只保留本次写入的统计结果
a 模式（追加写入）：打开文件时，指针移动到文件末尾。保留原有内容，本次写入的数据会直接接在旧数据后面。适合做日志记录
写入内容：根据实验要求，需统计出有效的行数、无效的行数，求有效记录的平均分和及格率，再通过 f-string 格式化输出写进文本中

捕获 FileNotFoundError触发原因：当尝试使用 open() 函数读取或写入一个路径错误或根本不存在的文件时，操作系统找不到对应的文件，就会触发 FileNotFoundError
捕获与输出：使用 try...except FileNotFoundError: 包裹读取代码。捕获到异常后，通常会输出一段友好的说明（如“文件未找到，请检查路径是否正确”），从而避免程序直接崩溃

实验5：类、继承与对象复制

```python
import copy

#5.1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name},score={self.score},passed={self.is_passed()})"

valid_records = [{"name":"张三","score":88},{"name":"李四","score":55},{"name":"王五","score":76}]

#5.2创建对象
for r in valid_records:
    s = Student(r["name"], r["score"])
    print(s)
#5.3 GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        base_str = super().__str__()
        return base_str + f", research_topic={self.research_topic}"
g = GraduateStudent("LiHua",88,"AI")
print(g)
#5.4深浅拷贝对比
origin = {"name": "Alice", "scores": [85, 90]}
#直接赋值
ref = origin
ref["scores"][0] = 999
print("直接赋值后 origin:", origin)
#复原
origin = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(origin)
shallow["scores"][0]= 777
print("copy.copy浅拷贝 origin:", origin)
#复原
origin = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(origin)
deep["scores"][0] = 111
print("deepcopy深拷贝 origin:", origin)
```
