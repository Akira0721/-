欧⼏⾥得算法

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))

# 负数测试
print("负数gcd(-12,-8):", gcd(-12,-8))
```

会返回 -4。因为 Python 中取模运算（%）的结果符号跟随除数，导致最后算出来的是负数。
没有意义。 数学上最大公约数（GCD）默认是正数。Python 官方自带的 math.gcd(-12, -8) 返回的也是正数 4。
在函数最前面加一句 m, n = abs(m), abs(n) 把负数转正就行。

近似欧拉数e

```python
# 1. euler_limit (伯努利极限 (1+1/x)^x )
def euler_limit(n):
    return (1 + 1/n)**n

# 2. euler_infinite_sum 前n项和；0!=1
def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k==0:
            fact =1
        else:
            fact *=k
        total += 1.0 / fact
    return total

#3. euler_approx，误差小于ϵ
def euler_approx(eps):
    true_e = math.exp(1)
    total =0.0
    fact =1
    k=0
    while abs(true_e - total) >= eps:
        if k==0:
            fact =1
        else:
            fact *=k
        total += 1.0/fact
        k +=1
    return total

#4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val:.6f}")


# 测试
print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("真实e", math.exp(1))
print("euler_approx(1e-4)", euler_approx(1e-4))
print_euler_sum_table(6)
```

测试整数属性

```python
#1 is_power_of_2，只用除法和取模
def is_power_of_2(n):
    while n >1:
        if n %2 !=0:
            return False
        n = n//2
    return True

#2 is_power(b,n) n是否是b的幂
def is_power(b,n):
    while n >1:
        if n % b !=0:
            return False
        n = n //b
    return True

print(is_power_of_2(8))
print(is_power_of_2(9))
print(is_power(3,81))
print(is_power(3,80))
```
